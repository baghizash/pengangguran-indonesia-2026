```python
import pandas as pd
import matplotlib.pyplot as plt
from pathlib import Path
D = Path('data/compiled')
pd.set_option('display.float_format', '{:.2f}'.format)
```

## 1. Muat data kompilasi

```python
nas = pd.read_csv(D/'tpt_nasional_series.csv')
pend = pd.read_csv(D/'tpt_pendidikan.csv')
umur = pd.read_csv(D/'tpt_umur.csv')
prov = pd.read_csv(D/'tpt_provinsi.csv')
dist = pd.read_csv(D/'distribusi_penganggur_pendidikan.csv')
print('nasional:', nas.shape, '| pendidikan:', pend.shape,
      '| umur:', umur.shape, '| provinsi:', prov.shape)
```

## 2. Tren TPT nasional: turun perlahan

```python
nas[['periode','tpt_persen','penganggur_jt','upah_buruh_jt']]
```

## 3. Paradoks pendidikan (Mei 2026)

```python
p = pend.set_index('pendidikan')[['feb_2026','mei_2026']]
p['gap_vs_sd_pp'] = p['mei_2026'] - p.loc['SD ke bawah','mei_2026']
p['ratio_vs_nasional'] = p['mei_2026'] / 4.65
p.sort_values('mei_2026', ascending=False)
```

## 4. Pengangguran muda meledak saat nasional turun

```python
umur.set_index('kelompok_umur')
muda = umur.set_index('kelompok_umur').loc['15-24 tahun']
print('15-24 thn Mei 2026:', muda['mei_2026'], '%')
print('Naik Feb->Mei 2026:', round(muda['mei_2026']-muda['feb_2026'],2), 'pp')
print('Rasio vs nasional:', round(muda['mei_2026']/4.65,2), 'x')
```

## 5. Jurang provinsi (Mei 2026)

```python
pm = prov[['provinsi','mei_2026']].dropna().sort_values('mei_2026', ascending=False)
pm.head(10)
print('Papua vs Bali:', round(7.08/1.52,2), 'x lipat')
```

## 6. 7,22 juta penganggur: siapa mereka?

```python
dist['juta_orang'] = (7.22 * dist['porsi_persen'] / 100).round(4)
dist.sort_values('porsi_persen', ascending=False)
print('Total porsi:', dist['porsi_persen'].sum().round(2), '%')
```
